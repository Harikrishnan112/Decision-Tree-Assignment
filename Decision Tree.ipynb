```
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree

data = pd.read_csv("/content/Decision_Tree.csv")
data

labelencoder=LabelEncoder()
data["Temperature"]=labelencoder.fit_transform(data["Temperature"])
data["Vibration"]=labelencoder.fit_transform(data["Vibration"])
data["Failure"]=labelencoder.fit_transform(data["Failure"])

X = data[["Temperature","Vibration"]]
Y = data["Failure"]

model = DecisionTreeClassifier(criterion="entropy")
model.fit(X,Y)

sample=[[2,1]]

prediction=model.predict(sample)
prediction

if prediction==0:
  print("The machine will have failure")
else:
  print("The machine will not have failure")

plt.figure(figsize=(10,15))
tree.plot_tree(
     model,
     feature_names=["Temperature","Vibration"],
     class_names=["Failure","No Failure"],

 )
plt.title("Decision Tree")
plt.show
```
